```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('placement.csv')
df.head()
```

```python
class LR:
  def __init__(self):
    self.m = None
    self.b = None

  def fit(self, X_train, Y_train):
    num = 0
    den = 0
    # Get scalar means for X_train and Y_train
    X_mean = X_train.mean().values[0]
    Y_mean = Y_train.mean()

    for i in range(X_train.shape[0]):
      # Access scalar values for X_train and Y_train by integer location
      X_val = X_train.iloc[i].values[0]
      Y_val = Y_train.iloc[i]

      num = num + ((X_val - X_mean) * (Y_val - Y_mean))
      den = den + ((X_val - X_mean) * (X_val - X_mean))

    self.m = num / den
    self.b = Y_mean - (self.m * X_mean)
    print(self.m)
    print(self.b)

  def predict(self, X_test):
    print(self.m * X_test + self.b)
```

```python
X = df.iloc[:,0:1]
Y = df.iloc[:,-1]
```

```python
X
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=2)
```

```python
lr = LR()
```

```python
lr.fit(X_train,Y_train)
```

```python
lr.predict(X_test)
```
